```python
#import libraries

from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from headroom import HeadroomConfig, HeadroomMode
from headroom.integrations import HeadroomChatModel
from langchain_huggingface.llms import HuggingFacePipeline
from headroom.integrations import get_tool_metrics
from transformers import pipeline
import os
```

```python
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1-0528",
    huggingfacehub_api_token="hf_token_placeholder",
    provider="auto",
)

chat_model = ChatHuggingFace(llm=llm)

long_content = "Hello!"

response = chat_model.invoke(long_content)
print(response.content)
print(response)
```

```python
    # --- Headroom-wrapped version ---

optimized_llm = HeadroomChatModel(
    chat_model,
)

optimized_response = optimized_llm.invoke(long_content)
print(optimized_response.content)
print(optimized_response)

result = optimized_llm.get_metrics()
print(result)
```
